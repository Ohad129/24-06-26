```python
# 6. Replace & map: Create a copy of the DataFrame.
# In the copy, replace Female with F and Male with M using .replace().
# Then solve the same change with .map()

#replace
df_copy = df.copy()
df_copy['Gender'] = df_copy['Gender'].replace(['Female', 'Male'], ['F', 'M'])
display(df_copy)

#map
df_copy = df.copy()
gender_map = {'Female':'F', 'Male':'M'}
df_copy['Gender'] = df_copy['Gender'].map(gender_map)
df_copy
```

```python
# 11. Missing data: Use the DataFrame below and implement the requested cleaning operations

df_missing = pd.DataFrame({
    'A': [None, 2, 3, None, 5, None],
    'B': [10, None, None, 40, 50, None],
    'C': [100, None, 300, 400, None, None]
})
df_missing
```

```python
# 11-1. Eliminate any row in the DataFrame that contains at least one NaN value

display(df_missing.dropna())
```

```python
# 11-2. Modify the previous task and eliminate only rows that have NaN values across all columns

display(df_missing.dropna(how='all'))
```

```python
# 11-3. Eliminate rows that have at least one NaN in columns A and C only

display(df_missing.dropna(subset=['A', 'C']))
```

```python
# 11-4. Replace all NaN values with a specified value of your choice

df_missing['A'] = df_missing['A'].fillna(0)
df_missing['B'] = df_missing['B'].fillna(1)
df_missing['C'] = df_missing['C'].fillna(2)

df_missing
```
